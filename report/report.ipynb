## 1.2.3. Exact solution

The two stationary solutions to the wave equations are real and imaginary components of the waves

$$u(t, x, y) = e^{\imath (k_x x + k_y y - \omega t)} \tag{1.6}$$

where the imaginary unit $\imath = \sqrt{-1}$. Show that Eq. $(1.6)$ satisfies the wave equation.


### Solution

We consider the 2D wave equation:

$$\frac{\partial^2 u}{\partial t^2} = c^2 \left( \frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2} \right).$$

We are given the proposed exact wave solution $(1.6)$.

1. Compute the second time derivative:

   $$\frac{\partial u}{\partial t} = -\imath \omega \cdot e^{\imath (k_x x + k_y y - \omega t)}$$

   $$\frac{\partial^2 u}{\partial t^2} = (-\imath \omega)^2 e^{\imath (k_x x + k_y y - \omega t)} = \imath^2 \omega^2 u(t,x,y) = -\omega^2 u(t,x,y)$$

2. Compute the second spatial derivatives:

   $$\frac{\partial^2 u}{\partial x^2} = (\imath k_x)^2 u(t,x,y) = -k_x^2 u(t,x,y)$$

   $$\frac{\partial^2 u}{\partial y^2} = (\imath k_y)^2 u(t,x,y) = -k_y^2 u(t,x,y)$$

3. Substitute into the wave equation:

   $$-\omega^2 u(t,x,y) = c^2 \left( -k_x^2 u(t,x,y) - k_y^2 u(t,x,y) \right)$$

4. Dispersion Relation:

   Canceling $-u(t,x,y)$ from both sides yields the continuous dispersion relation:
   $$\omega^2 = c^2 (k_x^2 + k_y^2) \implies \omega = c \sqrt{k_x^2 + k_y^2}$$

Thus, Eq. $(1.6)$ satisfies the wave equation provided that $\omega = c \sqrt{k_x^2 + k_y^2}$.


## 1.2.4. Dispersion coefficient

Assume that $m_x = m_y$ such that $k_x = k_y = k$. A discrete version of Eq. $(1.6)$ will then read

$$u_{ij}^n = e^{\imath (kh(i+j) - \tilde{\omega} n \Delta t)} \tag{1.7}$$

where $\tilde{\omega}$ is a numerical dispersion coefficient, i.e., the numerical approximation of the exact $\omega$. Insert Eq. $(1.7)$ into the discretized Eq. $(1.3)$ and show that for CFL number $C = 1/\sqrt{2}$ we get $\tilde{\omega} = \omega$.


### Solution

Assuming $m_x = m_y$, we have $k_x = k_y = k$. The discrete wave solution is given by

$$u_{ij}^n = e^{\imath (kh(i+j) - \tilde{\omega} n \Delta t)} \tag{1.7}$$

where $h = \Delta x = \Delta y$.

1. Discretized Wave Equation:

   Using equal spatial mesh spacing $h$ and time step $\Delta t$, Eq. $(1.3)$ reads as
   $$\frac{u_{i,j}^{n+1} - 2u_{i,j}^n + u_{i,j}^{n-1}}{\Delta t^2} = c^2 \left( \frac{u_{i+1,j}^n - 2u_{i,j}^n + u_{i-1,j}^n}{h^2} + \frac{u_{i,j+1}^n - 2u_{i,j}^n + u_{i,j-1}^n}{h^2} \right).$$

   Introducing the CFL number $C = \frac{c \Delta t}{h}$, we rearrange the stencil as
   $$u_{i,j}^{n+1} - 2u_{i,j}^n + u_{i,j}^{n-1} = C^2 \left( u_{i+1,j}^n + u_{i-1,j}^n + u_{i,j+1}^n + u_{i,j-1}^n - 4u_{i,j}^n \right).$$

2. Substitute $u_{ij}^n$ into the stencil:

   Dividing all terms by $u_{ij}^n = e^{\imath (kh(i+j) - \tilde{\omega}n\Delta t)}$:
   - Time terms:
     $$\frac{u_{ij}^{n+1}}{u_{ij}^n} = e^{-\imath \tilde{\omega} \Delta t}, \qquad \frac{u_{ij}^{n-1}}{u_{ij}^n} = e^{\imath \tilde{\omega} \Delta t}$$
     $$e^{-\imath \tilde{\omega} \Delta t} - 2 + e^{\imath \tilde{\omega} \Delta t} = 2 \cos(\tilde{\omega} \Delta t) - 2 = -4 \sin^2\left(\frac{\tilde{\omega} \Delta t}{2}\right)$$
   - Spatial terms ($x$-direction):
     $$\frac{u_{i+1,j}^n}{u_{ij}^n} = e^{\imath k h}, \qquad \frac{u_{i-1,j}^n}{u_{ij}^n} = e^{-\imath k h}$$
     $$e^{\imath k h} - 2 + e^{-\imath k h} = 2 \cos(kh) - 2 = -4 \sin^2\left(\frac{kh}{2}\right)$$

   - Spatial terms ($y$-direction):
     Since $k_x = k_y = k$, the $y$-direction yields the exact same term:
     $$-4 \sin^2\left(\frac{kh}{2}\right)$$

3. Simplify the algebraic system:

   Combining the terms:
   $$-4 \sin^2\left(\frac{\tilde{\omega} \Delta t}{2}\right) = C^2 \left( -4 \sin^2\left(\frac{kh}{2}\right) - 4 \sin^2\left(\frac{kh}{2}\right) \right)$$

   Dividing by $-4$:
   $$\sin^2\left(\frac{\tilde{\omega} \Delta t}{2}\right) = 2 C^2 \sin^2\left(\frac{kh}{2}\right)$$

   Taking the square root:
   $$\sin\left(\frac{\tilde{\omega} \Delta t}{2}\right) = \sqrt{2} C \sin\left(\frac{kh}{2}\right)$$

4. Apply the condition $C = 1/\sqrt{2}$:

   Substituting $C = \frac{1}{\sqrt{2}}$:
   $$\sin\left(\frac{\tilde{\omega} \Delta t}{2}\right) = \sqrt{2} \left(\frac{1}{\sqrt{2}}\right) \sin\left(\frac{kh}{2}\right) = \sin\left(\frac{kh}{2}\right)$$

   Equating the arguments:
   $$\frac{\tilde{\omega} \Delta t}{2} = \frac{kh}{2} \implies \tilde{\omega} = k \frac{h}{\Delta t}$$

5. Compare with exact frequency $\omega$:
   - Exact frequency: For $k_x = k_y = k$, the exact frequency from Section 1.2.3 is:
     $$\omega = c \sqrt{k^2 + k^2} = c k \sqrt{2}$$

   - Numerical frequency: From $C = \frac{c \Delta t}{h} = \frac{1}{\sqrt{2}}$, we have $\frac{h}{\Delta t} = c \sqrt{2}$. Substituting this into $\tilde{\omega}$:
     $$\tilde{\omega} = k \cdot (c \sqrt{2}) = c k \sqrt{2}$$

Therefore, for $C = 1/\sqrt{2}$, we obtain:

$$\tilde{\omega} = \omega.$$
